```python
!pip install sentence-transformers -q

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

#load model
model = SentenceTransformer('all-MiniLM-L6-v2')

#FAQs
faq_questions = [
    "what is AI",
    "what is machine learning",
    "How does Ddeep Learing work",
    "what is Python used for?"
]

faq_answers = [
    "AI enables machines to mimic human intelligence.",
    "Machine Learining allows systems to learn from data.",
    "Deep Learning uses neural networks with many layers.",
    "Python is widely usen in Aoi, ML, and web devlopment."
]

query = input("Ask your question: ")

#What is DL?

# Embeddings
faq_embeddings = model.encode(faq_questions)
query_embedding = model.encode([query])

# Similarity
scores = cosine_similarity(
    query_embedding,
    faq_embeddings
)

print(scores)

# Best answer
best_index = np.argmax(scores)

print("\nMost Relevant FAQ:")
print(faq_questions[best_index])

print("\nAnswer:")
print(faq_answers[best_index])
```

```python
# Safely delete the overridden 'input' variable if it exists in the global namespace
import sys
if 'input' in globals() and isinstance(globals()['input'], str):
    del globals()['input']

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')
text1 = input("Enter text 1: ")
text2 = input("Enter text 2: ")

text1_embeddings = model.encode(text1)
text2_embeddings = model.encode(text2)

scores = cosine_similarity(
    [text1_embeddings],
    [text2_embeddings]
)

print(
    "Text 1:", text1,
    "\nText 2:", text2,
    "\nScore:", scores[0][0]
)

if scores[0][0] > 0.8:
    print("The two texts are very similar.")
elif scores[0][0] < 0.4:
    print("The two texts are not at all similar.")
else:
    print("The two texts are somewhat similar.")
```
